```python

# quick function to change c to f. math is (c * 9/5) + 32
def convert(celsius):
    # TODO: maybe handle invalid inputs here later? keeping it simple for now
    ans = (celsius * 9 / 5) + 32
    return ans

# getting threshold from user
raw_input_thresh = input("Enter the temperature threshold: ")
threshold = float(raw_input_thresh)

# lets do this 3 times as requested, could probably use a while loop but range is fine
for i in range(3):
    # debug print just to see which iteration we are on, remove later???
    # print(f"Loop index: {i}")

    temp_str = input("Enter temperature in Celsius: ")
    celsius = float(temp_str)

    fahrenheit = convert(celsius)

    print("Temperature in Fahrenheit:", fahrenheit)

    # checking thresholds now
    if celsius > threshold:
        print("Temperature is above the threshold")
    else:
        if celsius < threshold:
            print("Temperature is below the threshold")
        else:
            # must be equal then
            print("Temperature is equal to the threshold")

    print() # extra space for readability
```
